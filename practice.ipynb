## Nested If

```python
a=int(input("Enter a value for a: "))
b=int(input("Enter a value for b: "))
c=int(input("Enter a value for c: "))
if (a>=b):
    if (a>=c):
        greatest=a
    else:
        greatest=c
else:
    if (b>=c):
        greatest=b
    else:
        greatest=c
print(str(greatest)+" is the greatest number.")
```
