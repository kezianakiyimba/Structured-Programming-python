```python
test1=int(input("Enter the mark: "))
test2=int(input("Enter the mark: "))
test3=int(input("Enter the mark: "))
test4=int(input("Enter the mark: "))
test5=int(input("Enter the mark: "))
average=(test1+test2+test3+test4+test5)/5
if average>=80:
    print("You are an outstanding student.")
if average>=70 and average<80:
    print("You are a good student.")
if average>=60 and average<70:
    print("You are an average student.")
if average>=50 and average<60:
    print("You are a below-average student.")
if average>=40 and average<50:
    print("You are a poor student.")
if average<40:
    print("You need extra ordinary efforts.")
print(f"Your average mark is: {average}")
```
