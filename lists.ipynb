```python
marks = []
n=int(input("How many marks? "))

for i in range(n):
    value=int(input("Enter mark: "))
    marks.append(value)

if len(marks)>0:
    average = sum(marks) / len(marks)
    print("Average:",average)
    print("Highest:",max(marks))
    print("Lowest:",min(marks))
else:
    print("No marks entered.")
```

write a function that takes a list of items and returns how many are above 50.

```python
numbers = []
def count_those_above_50(numbers):
    count = 0
    for number in numbers:
        if number > 50:
            count = count + 1 #here I think you can also use +=1
    return count

n = int(input("How many numbers are you using? "))
for i in range(n):
    value = int(input("Enter a number: "))
    numbers.append(value)

print(count_those_above_50(numbers))
```

Write a small menu (Add/View/Exit) using match case.
Use a while true loop for the menu.

```python
items = []

while True:
    choice = input("Please select(Add/View/Exit): ") 

    match choice:
        case "Add":
            item = input("What item would you like to add?: ")
            items.append(item)
            print("That item has been added.")

        case "View":
            if items:
                print(items)
            else:
                print("The list is empty!")

        case "Exit":
            print("Have a nice day!")
            break

        case _:
            print("Please choose Add, View, or Exit.")
```
